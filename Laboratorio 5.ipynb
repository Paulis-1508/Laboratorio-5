**Punto 4**

```python
conteo_calidad = df['Data quality'].value_counts()

df_confiable = df[df['Data quality'].isin(['acceptable', 'high'])].copy()

n_total = len(df)
n_confiable = len(df_confiable)
porcentaje = (n_confiable / n_total) * 100

print(f"Porcentaje: {porcentaje:.2f}%")
```

**Punto 5**

```python
longevidad = df_confiable["Maximum longevity (yrs)"]

media = longevidad.mean()
mediana = longevidad.median()
moda = longevidad.mode()

print("Media:", media)
print("Mediana:", mediana)
print("Moda:", moda)
```

La media de longevidad es de aproximadamente 18.95 años, mientras que la mediana es de 15 años y la moda es de 10 años. La media y la mediana sí son diferentes, ya que existe una diferencia de aproximadamente 3.95 años. La media es mayor que la mediana, lo que indica que algunos valores altos de longevidad pueden estar aumentando el promedio. Esto sugiere que la distribución de la longevidad presenta cierto sesgo hacia valores altos.

**Punto 6**

```python
longevidad = df_confiable['Maximum longevity (yrs)']

rango = longevidad.max() - longevidad.min()
varianza = longevidad.var()
desv_est = longevidad.std()

rango, varianza, desv_est
```

Estos estadísticos muestran que la longevidad varía mucho entre especies: el rango refleja la enorme distancia entre la especie más longeva y la más corta, pero depende solo de dos casos extremos, mientras que la varianza y la desviación estándar confirman que esa dispersión es alta en todo el conjunto y no un artefacto de uno o dos valores; en conjunto indican que no existe un "ritmo de vida" único en el reino animal, sino una distribución muy dispersa.

**Punto 7**

Para representar la longevidad típica de una especie, la medida mas adecuada seria la mediana, ya que cuando existen valores extremos que hacen que la media se aleje del como del centro de los datos, la mediana es menos sensible a valores muy muy altos o muy muy bajos, por lo que resulta más representativa, cuando la distribución de la longevidad  de una especie está sesgada a esos valores extremos que se pueden presentar.

**Punto 8**

```python
longevidad_promedio = (
    df_confiable.groupby("Class")["Maximum longevity (yrs)"].mean().sort_values(ascending=False)
)

print(longevidad_promedio.round(2))
```

Chondrichthyes presenta la mayor longevidad promedio

Solo coincide en parte con la intuición. Que mamíferos y reptiles figuren entre los más longevos no sorprende, ya que incluyen ballenas, elefantes y tortugas gigantes y animales grandes. Pero que Chondrichthyes encabece la lista es menos intuitivo, porque uno tiende a pensar en tiburones como depredadores de vida "rápida". El resultado está dominado por unas pocas especies de aguas profundas y frías, lo que también se refleja en que esta clase tiene la desviación estándar más alta y muy pocas especie frente a Aves o Mammalia. Por eso el promedio no representa a un tiburón "típico", sino que está inflado por valores extremos, y conviene revisarlo junto con la mediana (que para Chondrichthyes baja a 19.25, mucho más cercana a Reptilia) antes de tomarlo como un patrón biológico confiable.

**Punto 9**

```python
comparacion_origen = (
    df_confiable
    .groupby("Specimen origin")["Maximum longevity (yrs)"]
    .agg(["mean", "std"])
)
print(comparacion_origen)
```

La mayor longevidad promedio se presenta en los animales de cautiverio, con aproximadamente 20.47 años. En comparación, los animales con origen desconocido tienen un promedio de 17.87 años, mientras que los animales en estado silvestre presentan el menor promedio, con 17.57 años. En cuanto a la variabilidad, el grupo silvestre tiene la desviación estándar más alta 19.58 años. En cambio, el cautiverio presenta una desviación estándar menor 14.08 años. Una posible explicación de la mayor longevidad en cautiverio es que los animales pueden tener alimentación controlada, atención veterinaria y menor exposición a depredadores. Sin embargo, estos resultados no permiten afirmar que el cautiverio sea directamente la causa de una mayor longevidad.
